---

###  **Business Idea**

Modern sports organizations collect massive amounts of performance and training data, yet most insights remain siloed — physical performance metrics are rarely analyzed alongside mental readiness indicators.
**Athlete360** aims to bridge that gap by developing a data-driven system that integrates player performance statistics, workload metrics, and psychological sentiment data to provide a holistic view of athlete readiness.

This system will help **coaches, analysts, and athletes** make more informed decisions about:

* Player rotation and rest,
* Training focus and recovery strategies,
* Injury prevention,
* And performance optimization under pressure.

---

### **Vision**

> **To create an intelligent analytics platform that merges physical performance and mental state indicators, enabling smarter, safer, and more personalized coaching decisions.**

Athlete360 envisions an ecosystem where **AI supports human intuition** — empowering athletes to perform better while reducing risk of overtraining or injury.

### **Data Quality Checks**

| **Issue**        | **Explanation**                                                     | **Next Step / Solution**                                      |
| ---------------- | ------------------------------------------------------------------- | ------------------------------------------------------------- |
| Missing values   | Common in rolling and fatigue-related columns early in each season. | Impute using player averages or drop first few games.         |
| Player imbalance | Some players have fewer games recorded (e.g., rookies or injuries). | Focus on players with consistent participation (≥ 100 games). |
| Data scaling     | Metrics have different scales (e.g., PTS vs FG_PCT).                | Standardize before modeling or clustering.                    |

### Data Preprocessing

```python
predictive_features = [
    'minutes_played', 'days_since_last_game', 'back_to_back_games', 'games_last_7', 'games_last_30',
    'rolling_points_mean', 'rolling_points_std', 
    'rolling_assists_mean', 'rolling_assists_std',
    'rolling_rebounds_mean', 'rolling_rebounds_std',
    'rolling_field_goal_pct_mean', 'rolling_field_goal_pct_std',
    'rolling_3pt_pct_mean', 'rolling_3pt_pct_std',
    'rolling_free_throws_mean', 'rolling_free_throws_std',
    'rolling_turnovers_mean', 'rolling_turnovers_std',
    'rolling_minutes_mean', 'rolling_minutes_std',
    'last_vs_rolling_points_diff', 'hot_streak_flag', 'hot_streak_count',
    'fatigue_index', 'gap_days',
    'opponent_avg_blocks', 'opponent_avg_steals', 'opponent_def_rating',
    'opponent_pace', 'opponent_avg_points_allowed',
    'opponent_field_goals_pct_allowed', 'opponent_3pt_pct_allowed',
    'opponent_points_paint_allowed', 'cluster'
]
```

```python
# define target
target = "points" 

# Determine latest season
seasons_sorted = sorted(data["season"].unique())
test_season = seasons_sorted[-1]
print(f"Using {test_season} as test season")

# Split into train/test
is_test = data["season"] == test_season
is_train = ~is_test

X_train = data.loc[is_train, predictive_features]
y_train = data.loc[is_train, target]
X_test  = data.loc[is_test, predictive_features]
y_test  = data.loc[is_test, target]
print("Train size:", X_train.shape, "| Test size:", X_test.shape)
```

**(i) Linear Regression-Baseline Model**

```python
#Scale
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform (X_test)

# Initialize model
lr = LinearRegression()
lr.fit(X_train_scaled, y_train)

# Predictions
y_pred = lr.predict(X_test_scaled)

# Evaluation
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Linear Regression MAE: {mae:.2f}")
print(f"Linear Regression R²: {r2:.2f}")
```

**(ii) Random Forest Regressor**

```python
# Initialize baseline RF
rf_baseline = RandomForestRegressor(random_state=42, n_jobs=-1)

# Train
rf_baseline.fit(X_train, y_train)

#Predict
y_pred_base = rf_baseline.predict(X_test)

# Evaluate
base_mae = mean_absolute_error(y_test, y_pred_base)
base_r2 = r2_score(y_test, y_pred_base)

print(f"Baseline Random Forest MAE: {base_mae:.2f}")
print(f"Baseline Random Forest R²: {base_r2:.2f}")
```

```python
top_n = 6   
cols = 3    
rolling_window = 5  

# Identify top scorers
top_players = (
    data.groupby("player_name")["points"]
    .mean()
    .sort_values(ascending=False)
    .head(top_n)
    .index.tolist()
)

print(f"Top {top_n} players for visualization:", top_players)


# Attach predictions to test data
test_with_preds = data.loc[data["season"] == test_season].copy()
test_with_preds["predicted_points"] = y_pred

rows = int(np.ceil(top_n / cols))
fig, axes = plt.subplots(rows, cols, figsize=(18, 5 * rows), sharex=False)
axes = axes.flatten()

for i, player_name in enumerate(top_players):
    player_data = test_with_preds[test_with_preds["player_name"] == player_name].copy()
    player_data = player_data.sort_values("game_date")

    if player_data.empty:
        continue

    # Compute rolling average of actual points
    player_data["rolling_avg_points"] = player_data["points"].rolling(window=rolling_window).mean()

    ax = axes[i]
    ax.plot(player_data["game_date"], player_data["points"],
            label="Actual", marker="o", color="royalblue", alpha=0.7)
    ax.plot(player_data["game_date"], player_data["predicted_points"],
            label="Predicted", marker="x", linestyle="--", color="orange", alpha=0.8)
    ax.plot(player_data["game_date"], player_data["rolling_avg_points"],
            label=f"{rolling_window}-Game Avg", color="green", linewidth=2, alpha=0.7)

    ax.set_title(player_name, fontsize=12, fontweight="bold")
    ax.set_xlabel("Game Date")
    ax.set_ylabel("Points")
    ax.tick_params(axis='x', rotation=45)
    ax.grid(alpha=0.3)

# Clean Up
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

fig.suptitle(
    f"Top {top_n} Players — Actual vs Predicted Next-Game Points with {rolling_window}-Game Average ({test_season})",
    fontsize=16, y=1.03, fontweight="bold"
)
fig.tight_layout()
fig.legend(["Actual", "Predicted", f"{rolling_window}-Game Avg"],
           loc="upper right", bbox_to_anchor=(1.15, 1))
plt.show()
```
